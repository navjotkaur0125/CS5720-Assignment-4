```python
import re
import nltk
from nltk.stem import PorterStemmer

def nlp_preprocessing_pipeline(sentence):
    # Step 1: Tokenize sentence into words and punctuation
    tokens = re.findall(r"\w+|[^\w\s]", sentence)

    # Step 2: Define and remove common English stop words
    stopwords = {"are", "used", "in", "and", "the", "a", "an", "of", "to", "is", "it"}
    tokens_no_stopwords = [word for word in tokens if word.lower() not in stopwords]

    # Step 3: Apply stemming to reduce words to root form
    stemmer = PorterStemmer()
    stemmed_words = [stemmer.stem(word) for word in tokens_no_stopwords]

    return tokens, tokens_no_stopwords, stemmed_words

# Test Sentence from Home Assignment 4
test_sentence = "NLP techniques are used in virtual assistants like Alexa and Siri."
tokens, no_stopwords, stemmed = nlp_preprocessing_pipeline(test_sentence)

print("=== Task 1: NLP Preprocessing Pipeline Outputs ===")
print("Original Tokens:          ", tokens)
print("Tokens Without Stopwords: ", no_stopwords)
print("Stemmed Words:            ", stemmed)
```

```python
import numpy as np

def scaled_dot_product_attention(Q, K, V):
    d_k = Q.shape[-1]

    # 1 & 2. Compute Q K^T scaled by sqrt(d_k)
    scores = np.dot(Q, K.T) / np.sqrt(d_k)

    # 3. Softmax across the last axis
    exp_scores = np.exp(scores - np.max(scores, axis=-1, keepdims=True))
    attention_weights = exp_scores / np.sum(exp_scores, axis=-1, keepdims=True)

    # 4. Multiply by V
    output = np.dot(attention_weights, V)

    return attention_weights, output

# --- ASSIGNMENT 4 TEST INPUTS (Sequence length of 2) ---
Q = np.array([[1, 0, 1, 0],
              [0, 1, 0, 1]], dtype=float)

K = np.array([[1, 0, 1, 0],
              [0, 1, 0, 1]], dtype=float)

# V interpreted correctly as a 2x4 matrix of ranges 1-4 and 5-8
V = np.array([[1, 2, 3, 4],
              [5, 6, 7, 8]], dtype=float)

attn_weights, final_output = scaled_dot_product_attention(Q, K, V)

print("=== Task 2: Scaled Dot-Product Attention Outputs ===")
print("Attention Weights Matrix (Softmax):")
print(np.round(attn_weights, 4))

print("\nFinal Output Matrix:")
print(np.round(final_output, 4))
```
